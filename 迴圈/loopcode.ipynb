# 九九乘法表
1\*1=1, 1\*2=2, 1\*3=3, ...
2\*1=1, 2\*2=2, 2\*3=3, ...
3\*1=1, 3\*2=2, 3\*3=3, ...

```python
for j in range(1, 6):
 print(f'1*{j}={1*j}' ,end=',')
```

```python
for i in range(1, 6):
    for j in range(1, 6):
        print(f'{i}*{j}={i*j}', end=',')
    print()
```

# 練習 : 印出質數
質數:一個數字n，除了自己本，以及1以外，沒有任何樹可以整除他，n就是質數。

```python
n=273
i=2
if 273 % 2 == 0:
    print(f"{273}不是質數")
i=3
if 273 % 3 == 0:
    print(f"{273}不是質數")
i=4
if 273 % 4 == 0:
    print(f"{273}不是質數")
```

```python

n = 17
isPrime = True # 假設n是質數
for i in range(2, n):
    if n % i == 0:
        print(f"{n}不是質數,因為它可以被 {i}除盡")
        isPrime = False # 確定n不是質數
        break # 只要數字能被除盡，就跳離迴圈
if isPrime ==True: #判斷我們的假設是還成立
  print(f'{n}是質數')
```
