# 強制啟動 MouseInfo 視窗工具
透過顯式呼叫 tkinter 初始化與主迴圈，強制讓 PyAutoGUI 的 mouseInfo() 視窗在新版 Python 環境中彈出。

```python
import pyautogui as pyg
import tkinter as tk

print("正在強制啟動 MouseInfo 視窗，請稍候...")

try:
    # 先嘗試直接調用原版 mouseInfo
    pyg.mouseInfo()
except Exception as e:
    print(f"原版調用異常 ({e})，正在使用 tkinter 強制渲染視窗...")
    # 如果因為執行緒或 GUI 阻擋導致無法彈出，使用手動建立 tkinter 視窗來強制顯示 mouseInfo 介面
    root = tk.Tk()
    root.withdraw() # 隱藏主視窗
    # 直接呼叫 mouseInfo 內部的視窗建構
    from mouseInfo import MouseInfo
    app = MouseInfo()
    root.mainloop()
```
